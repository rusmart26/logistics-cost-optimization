```python
grupos_ruta= (orders_flete_complet.groupby(['orig_port','dest_port'])["shipping_cost"].agg(["mean","median","count"]).reset_index())

print(grupos_ruta)
```

```python
grupos_producto=(orders_flete_complet.groupby(["product_id"])["shipping_cost"].agg(["mean","median","count"]).reset_index())

grupos_producto['skew_ratio'] = grupos_producto['mean'] / grupos_producto['median']

grupos_producto_filter= grupos_producto[(grupos_producto["count"]>10)& (grupos_producto['mean']>50)]
grupos_producto_sorted=grupos_producto_filter.sort_values(by='skew_ratio', ascending=False)
print(grupos_producto_sorted)
print("count>10:", (grupos_producto['count']>10).sum())
print("mean>50:", (grupos_producto['mean']>50).sum())
```

```python
print(grupos_producto[grupos_producto['mean']>50].sort_values(by='mean', ascending=False))
```

```python
grupos_producto.to_csv("grupos_producto.csv", index=False, encoding='utf-8-sig')
print("¡Tabla exportada a CSV exitosamente!")
```

```python
tarifas_unicas = orders_flete_complet[['min_cost', 'rate']].drop_duplicates()

tarifas_unicas['threshold']= tarifas_unicas['min_cost']/tarifas_unicas['rate']
print(tarifas_unicas)
```

```python
weight_celinig= pd.merge(orders_flete_complet, tarifas_unicas, on=['min_cost','rate'], how="left")

print(weight_celinig)
```

```python
below_threshold = (weight_celinig['weight'] < weight_celinig['threshold']).sum()
total = len(weight_celinig)
print(f"{below_threshold} de {total} ({below_threshold/total*100:.1f}%)")
```
